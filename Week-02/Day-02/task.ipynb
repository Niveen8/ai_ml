 Step 1: Simulate 10,000 coin flips with NumPy and confirm the proportion of heads approaches 0.5.

```python
import numpy as np
```

```python
coin_fli = np.random.choice(["H","T"], size = 10000)
print(coin_fli)
heads = np.sum(coin_fli == "H")
print("heads = ",heads)
proportion = heads / 10000
print("Heads proportion =", proportion)
```

Coin Flip Simulation
The coin were flipped 10,000 times with NumPy random.
The fraction of Heads was close to 0.5, which is the probability of getting Heads in a toss with a fair coin, theoretically.
Demonstrating  that the experimental probability converges to the expected probability as the number of trials goes to infinity one.

Step 3: Given a small worked scenario, compute a conditional probability by hand and verify it with a
simulation.

```python
devices = np.array(["Abnormal_Fault"] * 30 +["Abnormal_Normal"] * 30 +["Normal"] * 40)
simulation = np.random.choice(devices, size=10000)
abnormal_devices = simulation[(simulation == "Abnormal_Fault") |(simulation == "Abnormal_Normal")]
fault_devices = np.sum(abnormal_devices == "Abnormal_Fault")
probability = fault_devices / len(abnormal_devices)
print("P(Fault | Abnormal) =", probability)
```

Result:
This example uses conditional probability in a medical engineering context.
Manual calculation yields the probability of a device fault conditioned on an abnormal signal of 0.5.
The simulation yielded a similar result thus confirming the mathematical calculation.
