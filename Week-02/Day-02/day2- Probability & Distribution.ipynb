# P(A | B) = P(A and B) / P(B)

```python
p_a_and_b = 0.1
p_b = 0.2
p_a_b = p_a_and_b / p_b
print("p(A | b) = ", p_a_b  )
```

```python
p_a_and_b = 0.18
p_b = 0.6
p_a_b = p_a_and_b / p_b
print("p(A | b) = ", p_a_b  )
```

------------------------------------------------

P(A | B) = ( P(B | A) × P(A) ) / P(B)


```python
p_b_a = 0.8
p_a = 0.3
p_b = 0.6
print("p(A | b) = ", ((p_b_a * p_a )/p_b))
```

```python
p_b_a = 0.7
p_a = 0.5
p_b = 0.8
p_a_b =  ((p_b_a * p_a ) / p_b)
print("p(A | b) = ",p_a_b )
```
