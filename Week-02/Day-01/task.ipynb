```python
import pandas as pd

columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]

df = pd.read_csv("iris.csv", names=columns)

df.head()
```

```python
import numpy as np

sepal_length_array = df["sepal_length"].to_numpy()

sepal_length_array[:10]
```


● Step 2: Compute its mean, median, mode, standard deviation, and IQR.

```python
from scipy import stats
```

```python


print(np.mean(sepal_length_array))
print( np.median(sepal_length_array))
print(stats.mode(sepal_length_array).mode)
print(np.std(sepal_length_array))
Q1 = np.percentile(sepal_length_array, 25)
Q3 = np.percentile(sepal_length_array, 75)
Q3 - Q1
print(Q3 - Q1)
```
