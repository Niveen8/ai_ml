1-mean:

```python
import numpy as np
arr = [5,6,11]
mean = np.mean(arr)
print(mean)
```

2-mode:

```python
import numpy as np
from scipy import stats
arr = [1,2,2,3]
mode = stats.mode(arr)
print(mode)
```

3-median:

```python
import numpy as np
arr = [1,2,3,4]
median = np.median(arr)
print(median)
```

1-range:

```python
import numpy as np
arr =[1,2,3,4,5]
maximum = max(arr)
minimum = min(arr)
range = maximum - minimum
print("max={},min={},range={}" .format(maximum,minimum,range))
```

```python
data = np.array([10, 12, 12, 13, 100])
print(np.std(data))
print(np.var(data)) 
q1, q3 = np.percentile(data, [25, 75])
print(q3 - q1) 
```
