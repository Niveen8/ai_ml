```python
import pandas as pd
import pandas as np
week_data = [1, 2, 3, 4, 5]
sales_data = [2, 4, 5, 4, 5]
df = pd.DataFrame({"Weeks":week_data ,
                   "Sales":sales_data})
```

```python

X = df['Weeks'].values
```

```python
Y = df['Sales'].values
```

```python
import numpy as np
import pandas as pd
mean_x = np.mean(X)
mean_y = np.mean(Y)
n = len(X)
numer = 0
denom = 0
for i in range(n):
    numer += (X[i] - mean_x) * (Y[i] - mean_y)
    denom += (X[i] - mean_x) ** 2
m = numer / denom
m
c = mean_y - (m * mean_x)
c
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_pred = c + (m * X)

print("MAE:", mean_absolute_error(Y, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(Y, y_pred)))
print("R2:", r2_score(Y, y_pred))
```

3.2 Why Accuracy Alone Is Misleading

Accuracy by itself is not informative while working with imbalanced datasets. 
For instance, if 95% of the customers are not churned, a model predicting all customers as "not churn" will get 95% accuracy, but it is not useful as it fails to identify the customers who will actually leave.
Hence, the classification model may be assessed through other evaluation metrics such as Precision, Recall, F1-score, and Confusion Matrix in order to have better insights on the model performance. 

---------------------------------------

3.3 The Confusion Matrix

A confusion matrix is a tool that visualizes the performance of a classification model on a set of test data for which true values are known [1].
It has four components:
1-TP (True Positive):the number of positive cases correctly predicted.

2-TN (True Negative):the number of negative cases correctly predicted.

3-FP (False Positive):the number of predicted positive cases which are not positive.

4-FN (False Negative):A case where the model fails to detect a positive case.

It's also used for deriving other metrics such as Accuracy, Precision, Recall, F1-score

------------------------------

3.4 Precision, Recall & F1


Precision, recall and F1-score can be used for evaluating the quality of classification models as well.
Precision: What proportion of cases predicted positive are actually positive.

Recall: How much of the true positive is the detector taking.

F1-score:A single measure of Precision and Recall.
What metric (in the sense of metrics) to count depends on the issue: for instance.

Disease detection ==>Recall is more important, because you just never want to miss any real ones.

Spam detection ==> precision is more important, you don't want to block legitimate emails.
