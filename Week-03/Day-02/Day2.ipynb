```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
week_data = [1,2,3,4,5]
sales_data = [2,4,5,4,5]
df = pd.DataFrame({"Weeks":week_data,"Sales":sales_data})
df
```

```python
X = df["Weeks"].values
```

```python
Y = df['Sales'].values
```

```python
mean_x = np.mean(X)
mean_y = np.mean(Y)
n = len(X)
numer = 0
denom = 0
for i in range(n):
    numer += (X[i] - mean_x) * (Y[i] * mean_y) 
m = numer /denom
c = mean_y - (m * mean_x)
print ( f"m :{m}" )
print ( f"c :{c}" )
```

```python
max_x = np.max(X) + 1
min_x = np.min(X) - 1
x = np.linspace(min_x,max_x)
y =(m * x) + c
plt.plot(x, y, color='#7D1E0C', label='Regression line', linestyle='--')
plt.scatter(X,Y,color='skyblue',label="Data Point")
plt.xlabel("Weeks")
plt.ylabel("Sales")
plt.title("LR Demo")
plt.grid(True)
plt.legend(loc='best')
```

2.1 What Linear Regression Does:

Linear Regression computes the value of a dependent variable d, by fitting a best fitting straight line through the data points of the observed d values.
The forecast is computed as:
Prediction = bias + ∑(feature × weight)Prediction=bias+∑(fr×wr) where r is the index of the features.
The model chooses the optimal weight and bias parameters such that the difference between what it predicts and the true values observed in data is as small as possible, during training. 

2.2 Training and Predicting:


```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

2.3 Interpreting Coefficients:


```python
print(model.coef_) 
print(model.intercept_)
```

2.4 Regression Metrics:

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
print(mean_absolute_error(y_test, predictions))
print(np.sqrt(mean_squared_error(y_test, predictions))) # RMSE
print(r2_score(y_test, predictions))
```
